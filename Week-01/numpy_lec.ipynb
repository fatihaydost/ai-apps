## Numerical Computing with NumPy
ML revolves around transforming data into numerical representations.
Arrays:
stores homogeneous data.
appending, extending, removing.
do not support broadcasting.

NumPy arrays:
allows you to mix data types.
Fourier transform, etc.
perform operations on arrays with different shapes.
achieves vectorization through broadcasting.

```python
import array
import numpy as np
import time

python_array = array.array('i', range(1000000))
numpy_array = np.arange(1000000)

# Time the sum operation for Python array
start_time = time.time()
sum(python_array)
end_time = time.time()
print('Python array sum time:', end_time - start_time)

# Time the sum operation for NumPy array
start_time = time.time()
np.sum(numpy_array)
end_time = time.time()
print('NumPy array sum time:', end_time - start_time)
```
