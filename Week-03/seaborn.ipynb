```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
# Read the CSV file into a DataFrame
df = pd.read_csv("iris.csv")
# Create a scatterplot using Seaborn
sns.scatterplot(x="PetalLength", y="PetalWidth", data=df, hue="Name")
# Show the plot
plt.show()
```

```python
# built-in tips dataset
tips = sns.load_dataset('tips')
```

```python
tips.head()
```

```python
sns.displot([tips['total_bill']])
```

```python
sns.displot([tips['total_bill']], kde=True) # Histogram
```

```python
sns.displot(x='total_bill', y='tip', data=tips) # Bivariate histogram/distribut
```

```python
sns.jointplot(x='total_bill', y='tip', data=tips) # Bivariate histogram/distrib
```

```python
sns.displot(x='total_bill', y='tip', data=tips, kind='hist') # Bivariate histog
```

```python
In [10]: sns.jointplot(x='total_bill', y='tip', data=tips, kind='hex') # Bivariate histo
```

```python
sns.rugplot(tips['total_bill'])
```
